```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split

# 1. Đọc dữ liệu
data = pd.read_csv("housing_data.csv")

# 2. Kỹ thuật tạo đặc trưng (Feature Engineering)
data["rooms_per_person"] = data["total_rooms"] / data["total_bedrooms"]

# 3. Tách biến đầu vào X và mục tiêu y
X = data[["total_rooms", "total_bedrooms", "rooms_per_person"]]
y = data["median_house_value"]

# 4. Chia tập train/test (80% train, 20% test)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# 5. Huấn luyện mô hình
model = LinearRegression()
model.fit(X_train, y_train)

# 6. Dự đoán & Đánh giá
y_pred = model.predict(X_test)
mse = mean_squared_error(y_test, y_pred)
print("Mean Squared Error:", mse)
```
