## Introduction: Machine Learning Project - Classifying Spam Emails

Trong notebook này, chúng ta sẽ giải quyết trọn vẹn bài toán phân loại Email rác (Spam Emails) trên bộ dữ liệu thực tế.
Tóm tắt ngắn gọn mục tiêu của bài toán là:   
Sử dụng dữ liệu nội dung email được cung cấp để phát triển một mô hình có thể dự đoán một email là "spam" (thư rác) hay "not spam" (ham - thư thường), sau đó giải thích kết quả để tìm ra những biến số/từ khóa nào có tính dự báo cao nhất. 

Đây là bài toán Học có giám sát (Supervised), Phân loại nhị phân (Binary Classification):   
Supervised problem: Chúng ta được cung cấp cả các đặc trưng (features) và nhãn mục tiêu (target).   
Binary Classification problem: Biến mục tiêu là biến phân loại nhị phân gồm 2 lớp: 0 (Not Spam / Ham) và 1 (Spam). 



## Mã giả (Pseudocode) & Giải thích Tham số / Đặc trưng
1. Mã giả quy trình thuật toán (Pseudocode)

ALGORITHM: 8-STEP MACHINE LEARNING PIPELINE FOR SPAM CLASSIFICATION

INPUT: Raw Email Dataset D = {(text_i, label_i)}

OUTPUT: Tuned Best Model (M_tuned), Feature Importance Chart, & Evaluation Report



BEGIN


 Bước 1 : Làm sạch dữ liệu 

 Load dataset D, xóa các dòng trống (null) hoặc trùng lặp (duplicates)

 Mã hóa nhãn (Label Encoding): 'ham' -> 0, 'spam' -> 1
 
 Tạo hàm làm sạch văn bản: Chuyển chữ thường, xóa thẻ HTML, URL, dấu câu, số và Stopwords




STEP 2: EXPLORATORY DATA ANALYSIS (EDA)

Thống kê tỷ lệ mất cân bằng nhãn (Spam vs Ham) -> Vẽ Pie/Bar Chart

Trực quan hóa các từ khóa xuất hiện nhiều nhất ở mỗi nhóm -> Vẽ WordCloud




STEP 3: FEATURE ENGINEERING AND SELECTION

Tạo đặc trưng phụ từ văn bản gốc: F_excl (số dấu '!'), F_curr (số dấu '$','£'), F_len (độ dài)

Vẽ biểu đồ tương quan/phân phối của đặc trưng phụ giữa Spam và Ham

Vector hóa văn bản sạch bằng TF-IDF (chọn lọc top 3000 từ quan trọng nhất)

Ghép (hstack) ma trận TF-IDF và đặc trưng phụ -> Chia tập Train (80%) và Test (20)


STEP 4: ESTABLISH BASELINE & COMPARE MODELS

Thiết lập Baseline Model: DummyClassifier (dự đoán theo nhãn chiếm đa số)

Huấn luyện 4 mô hình ML: Naive Bayes, Logistic Regression, SVM, Random Forest trên tập Train

Đánh giá Cross-Validation / Validation bằng chỉ số chính là F1-Score -> Vẽ biểu đồ so sánh

Chọn mô hình có F1-Score cao nhất làm M_best



STEP 5: HYPERPARAMETER TUNING ON THE BEST MODEL

Định nghĩa lưới tham số (Parameter Grid) cho M_best (ví dụ: hệ số C, class_weight)

Chạy GridSearchCV (5-fold Cross Validation, tối ưu hóa theo F1-Score) để tìm M_tuned

STEP 6: EVALUATE THE BEST MODEL ON TESTING SET

Dùng M_tuned dự đoán trên tập dữ liệu kiểm thử (X_test)

Tính Accuracy, Precision, Recall, F1-Score -> Vẽ Confusion Matrix và Đường cong ROC-AUC

STEP 7: INTERPRET MODEL RESULTS

Trích xuất trọng số (Coefficients / Feature Weights) của M_tuned
    
Vẽ biểu đồ Top 15 đặc trưng đẩy email thành SPAM và Top 15 đặc trưng giữ email là HAM

STEP 8: DRAW CONCLUSIONS & REPORT

Tổng hợp kết quả định lượng và chạy thử nghiệm dự đoán trên email thực tế
